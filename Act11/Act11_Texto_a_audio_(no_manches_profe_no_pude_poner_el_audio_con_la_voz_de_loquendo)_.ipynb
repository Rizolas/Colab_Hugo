```
import edge_tts
import asyncio
from IPython.display import Audio, display

# ✏️ Escribe tu texto aquí
texto = "Hola gente de youtube como estan hoy les vengo a mostrar como instalar photoshop gratis xdxd yo no lo descargo porque ya lo tengo"

# 🎙️ Voces disponibles en español:
# es-MX-DaliaNeural     → Mujer, ]México
# es-MX-JorgeNeural     → Hombre, México
# es-ES-ElviraNeural    → Mujer, España
# es-ES-AlvaroNeural    → Hombre, España
# es-AR-ElenaNeural     → Mujer, Argentina

voz = "es-ES-AlvaroNeural"
archivo_salida = "output_edge.mp3"

async def generar_audio():
    comunicar = edge_tts.Communicate(texto, voz)
    await comunicar.save(archivo_salida)


await generar_audio()


display(Audio(archivo_salida, autoplay=True))
print(f"✅ Audio generado: {archivo_salida}")
```
